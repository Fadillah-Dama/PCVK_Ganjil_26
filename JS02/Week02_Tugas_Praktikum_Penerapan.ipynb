```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

# Baca citra utama
img = cv.imread('ktp.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)


# TUGAS 1: Kombinasi Kanal Merah-Biru dan Hijau-Biru

# Kanal Merah-Biru (Kanal Hijau dinolkan -> indeks 1 = 0)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

# Kanal Hijau-Biru (Kanal Merah dinolkan -> indeks 0 = 0)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Kanal Merah - Biru (Tanpa Hijau)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Kanal Hijau - Biru (Tanpa Merah)')
plt.axis('off')
plt.show()


# TUGAS 2: Tiga Bentuk Pencerminan (Flip)

flip_v = cv.flip(img_rgb, 0)   # Vertikal (atas-bawah)
flip_h = cv.flip(img_rgb, 1)   # Horizontal (kiri-kanan)
flip_vh = cv.flip(img_rgb, -1) # Keduanya

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(flip_v)
plt.title('Pencerminan Vertikal (flipCode=0)')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_h)
plt.title('Pencerminan Horizontal (flipCode=1)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_vh)
plt.title('Pencerminan Vertikal & Horizontal (flipCode=-1)')
plt.axis('off')
plt.show()


# TUGAS 3: Rectangle, Circle, dan Text pada Foto Aktivitas

# Ganti dengan path foto aktivitas pribadi Anda
foto_aktivitas = img_rgb.copy()
h, w = foto_aktivitas.shape[:2]

# Menambahkan Rectangle (di sekitar wajah/badan)
cv.rectangle(foto_aktivitas, pt1=(int(w*0.65), int(h*0.15)), pt2=(int(w*1), int(h*0.70)), color=(255, 0, 0), thickness=3)

# Menambahkan Circle
cv.circle(foto_aktivitas, center=(int(w*0.82), int(h*0.4)), radius=int(min(w, h)*0.15), color=(0, 255, 0), thickness=3)

# Menambahkan Teks Nama
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(foto_aktivitas, text='Fadillah Dama', org=(int(w*0.1), int(h*0.97)),
           fontFace=font, fontScale=w/600, color=(255, 255, 0), thickness=2, lineType=cv.LINE_AA)

plt.figure(figsize=(8, 6))
plt.imshow(foto_aktivitas)
plt.title('Anotasi Geometri dan Teks')
plt.axis('off')
plt.show()
```
