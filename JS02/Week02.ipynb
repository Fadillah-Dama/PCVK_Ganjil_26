```python
import cv2 as cv
from google.colab.patches import cv2_imshow
import numpy as np
from skimage import io
import matplotlib.pyplot as plt
from google.colab import drive

# drive.mount('/content/drive')

# membaca gambar dari drive
img_path = '/content/drive/MyDrive/PCVK/ktp.jpg'
img = cv.imread(img_path)
cv2_imshow(img)
print("Resolusi gambar: ", img.shape[1], " x ", img.shape[0])

# menampilkan gambar dengan plt dan merubahnya ke format warna RGB
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()

# menampilkan grayscale
img_gray = cv.imread(img_path, cv.IMREAD_GRAYSCALE)
# tanpa cmap
plt.subplot(1, 2, 1)
plt.imshow(img_gray)
plt.title("tanpa cmap")
# dengan cmap
plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap='gray')
plt.title("dengan cmap")
plt.show()
# dengan cmap magma
plt.imshow(img_gray, cmap='magma')
plt.title("dengan cmap magma")
plt.show()

# melakukan resizing
img_resize = cv.resize(img, (200, 200))
plt.imshow(img_resize)
plt.show()

# melakukan flipping
img_flip = cv.flip(img_resize, 1)
plt.imshow(img_flip)
plt.show()
# menampilkan di canvas yang lebih besar
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(1,1,1)
ax.imshow(img_flip)
plt.show()

# membuat bentuk geometri 2D dari OpenCV
# membuat black image
black_img = np.zeros(shape=(530,530,3), dtype=np.uint8)
# menambahkan bentuk persegi panjang
cv.rectangle(black_img, pt1=(290,0), pt2=(520,130), color=(0,0,255), thickness=10)
# menambahkan bentuk persegi
cv.rectangle(black_img, pt1=(250,250), pt2=(350,350), color=(255,0,0), thickness=10)
# menambahkan bentuk lingkaran
cv.circle(black_img, center=(150,150), radius=(60), color=(0,255,0), thickness=10)
# menambahkan bentuk lingkaran penuh dan garis diagonal
cv.circle(black_img, center=(400,400), radius=50, color=(0,255,0), thickness=-1)
cv.line(black_img, pt1=(0,0), pt2=(530,530), color=(255,255,255), thickness=5)
# menambah teks PCVK is FUM
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='PCVK! is FUN', org=(10,500), fontFace=font, fontScale=2,
           color=(255,255,0), thickness=3, lineType=cv.LINE_AA)
plt.imshow(black_img)
plt.show()

# pembuatan black image dengan int32
black_img2 = np.zeros(shape=(530,530,3), dtype=np.int32)
# membuat bentuk layang layang
vertices = np.array([[150,350],[250,250],[450,350],[250,450]], dtype=np.int32)
pts = vertices.reshape((-1,1,2))
cv.polylines(black_img2, [pts], isClosed=True, color=(0,255,0), thickness=5)
plt.imshow(black_img2)
plt.show()








```
