```python
import cv2 as cv
import matplotlib.pyplot as plt
from google.colab import files

# Upload file KTM
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Baca citra dan konversi format ke RGB
ktm = cv.imread(filename)
ktm_rgb = cv.cvtColor(ktm, cv.COLOR_BGR2RGB)

# Salin gambar
ktm_masked = ktm_rgb.copy()

# Menutup tulisan "POLITEKNIK NEGERI MALANG"
cv.rectangle(ktm_masked, pt1=(170, 40), pt2=(610, 160), color=(10, 40, 100), thickness=-1)

# Menutup Logo dan Tulisan "BNI"
cv.rectangle(ktm_masked, pt1=(1000, 50), pt2=(1260, 170), color=(240, 90, 40), thickness=-1)

# Menutup tulisan Prodi "D-IV TEKNIK INFORMATIKA"
cv.rectangle(ktm_masked, pt1=(250, 720), pt2=(850, 770), color=(255, 200, 0), thickness=-1)

# Tampilkan Hasil
plt.figure(figsize=(12, 8))
plt.imshow(ktm_masked)
plt.title("Tugas Kelompok 3: Penutupan POLITEKNIK NEGERI MALANG, BNI, dan Prodi", fontsize=13)
plt.axis("off")
plt.show()
```
